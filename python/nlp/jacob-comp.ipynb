```python
import os
import nltk
from nltk.corpus import wordnet as wn
from nltk.probability import FreqDist
from nltk.tag import pos_tag
```

```python
#nltk.download('averaged_perceptron_tagger_eng')
#nltk.download('tagsets')
```

```python
with open("../../text-files/og-memo-scrape.txt") as file:
    # tokenize
    raw_text = file.read()
    tokens = raw_text.split()
    tokens = [w.lower() for w in tokens]
    # text object
    textObj = nltk.Text(tokens)

    #nltk.download('stopwords')
    stopwords = nltk.corpus.stopwords.words('english')
    filtered_tokens = [w for w in tokens if w.isalpha() and w not in stopwords]

    # set
    print(f"Count pre-set: {len(filtered_tokens)}")
    set_filtered_tokens = set(filtered_tokens)
    print(f"Count post-set: {len(set_filtered_tokens)}")

    # top 20
    fdist = FreqDist(filtered_tokens)
    most_common = fdist.most_common(10)
    print(fdist)
    print(most_common)

    tagged = pos_tag(tokens)
    print(f"Tagged: {tagged}")
    
    # concordance
    print(80*"=")
    print(f"Concordance of 'death':")
    textObj.concordance("death")

    for word, count in most_common:
        synset_count = len(wn.synsets(word))
        print(f"'{word}' has {synset_count} synsets")
```
